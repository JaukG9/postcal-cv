```python
import numpy as np
import pandas as pd
from PIL import Image
```

```python
#load an image
img = Image.open('hk-silksong.jpg')
img = img.convert('RGB')
img = np.asarray(img)

#print(img)
print(img[1][1][1])
print(len(img))
print(len(img[1]))
```

```python
reds = np.zeros((533,800))
greens = np.zeros((533,800))
blues = np.zeros((533,800))

for i in range(len(img)):
    for j in range(len(img[1])):
        reds[i][j] = img[i][j][0]

for i in range(len(img)):
    for j in range(len(img[1])):
        greens[i][j] = img[i][j][1]

for i in range(len(img)):
    for j in range(len(img[1])):
        blues[i][j] = img[i][j][2]

dr = pd.DataFrame(reds)
dg = pd.DataFrame(greens)
db = pd.DataFrame(blues)
```

```python
output_reds = "red_data.xlsx"
print(f"Writing data to {output_reds}...")

output_greens = "green_data.xlsx"
print(f"Writing data to {output_greens}...")

output_blues = "blue_data.xlsx"
print(f"Writing data to {output_blues}...")
```

```python
dr.to_excel(output_reds, index=False, engine='openpyxl')
dr.to_excel(output_greens, index=False, engine='openpyxl')
dr.to_excel(output_blues, index=False, engine='openpyxl')
print("Done! Excel files successfully created.")
```

```python
from PIL import Image
import matplotlib.pyplot as plt
import cv2

img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.imshow(img_gray, cmap='gray')
```

```python
sobelx=cv2.Sobel(src=img_gray, ddepth=cv2.CV_64F, dx=1, dy=0, ksize=5)
plt.imshow(sobelx)
```

```python
sobely=cv2.Sobel(src=img_gray, ddepth=cv2.CV_64F, dx=0, dy=1, ksize=5)
plt.imshow(sobely)
```

```python
sobelxy=cv2.Sobel(src=img_gray, ddepth=cv2.CV_64F, dx=1, dy=1, ksize=5)
plt.imshow(sobelxy)
```

```python
canny_edges=cv2.Canny(image=img_gray, threshold1=100, threshold2=120)
plt.imshow(canny_edges)
print(img.shape)
```
